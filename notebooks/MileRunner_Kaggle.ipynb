# 🏃 MileRunner on Kaggle — FREE, up to 12-hour sessions (no credit card)

Kaggle gives **30 GB RAM** and **12-hour** sessions for free (phone verification enables internet). That's the longest single free run without a card.

### Before you run
1. In the right sidebar: **Internet → On** (needs a one-time free phone verification under Settings).
2. **Accelerator → None (CPU)** is fine (GPU optional).
3. Then **Run All**. The last cell prints a public **`https://…gradio.live`** link — open it to watch the runners.

The session runs up to 12 h. Training auto-saves; to continue later, just start a new session and run again.
